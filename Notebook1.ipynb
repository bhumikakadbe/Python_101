```python
#DIFFERENT TYPES OF DATA TYPES--->String, Integer, Float, Boolean
name = "Bhumika Kadbe"
age = 20
cgpa =  8.6
vegan = False 


print(f" My name is {name}")
print(f" I am {age} years old")
print(f" My CGPA is {cgpa}")
print(f" I am a vegan: {vegan}")
```

```python
#LISTS---> Mubtale, indexed, ordered, allow duplicates
hobbies =["Doing Nothing", "Eating", "Criticizing", "laughing", "Teasing" ,"Sleeping", "Watching Movies", "Traveling", "Criticizing", "Dancing"]
print(f" My hobbies are: {hobbies}")
print(f"remove duplicates from hobbies: {set(hobbies)}")
```

```python
#TUPLE---> Immutable, ordered, indexed, allow duplicates
favorite_food = ("Briyani","Chicken","Pizza", "Burger", "Pasta", "Dosa", "Ice Cream")
print(f" My favorite foods are: {favorite_food}")
print(f"My most favorite food is: {favorite_food[0]}")
```

```python
#SET---> Mutable, unordered, unindexed, no duplicates
fav_no = {20, 15, 20.4, 1.9 , 22, 12 }
print(f" My favorite numbers are: {fav_no}")
```

```python
#DICTIONARY---> Mutable, unordered, no duplicates, accessed by key, key-value pair
dict={
    "Name" : "Bhumika Kadbe",
    "Age" : 20,
    "CGPA" : 8.6,
    "Subjects" : ["Python", "Data Science", "Machine Learning"],
    "three_sub" : True }
print(f" My name is {dict['Name']}")
print(f" I am {dict['Age']} years old")
print(f" My CGPA is {dict['CGPA']}")
print(f" I am studying {dict['Subjects']}")
print(f" I am studying three subjects: {dict['three_sub']}")
```
